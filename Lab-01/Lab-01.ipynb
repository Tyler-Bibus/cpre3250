We have already seen basic Python data types: our `int`, `float`, `str`, and `bool` variables used in the code examples above. Another commonly used data structure is the list, which allows us to store multiple values within a single variable. While a list may look similar to an array from other programming languages, it is not a true array; Python lists are more flexible and can even contain elements of different data types. Before moving on to additional data structures, let's review some common operations that can be performed on lists.

Here we see a powerful Python feature that allows us to check whether a value is present in a list without writing an explicit loop. The `in` and `not in` operators perform a membership test, making the code more concise, readable, and often more intuitive than manually iterating through each element of the list:

````python
x = 8
y = 13
primes_list = [2, 3, 5, 7, 11, 13, 17, 19, 23]

if (x not in primes_list):
    print(f"x = {x} is NOT present in primes list")
else:
    print(f"x = {x} is present in primes list")

if (y in primes_list):
    print(f"y = {y} is present in primes list")
else:   
    print(f"y = {y} is NOT present in primes list")
````

You can still use conventional bracket `[]` syntax to access individual elements in a list, like in traditional arrays. Python lists are 0-indexed, meaning the first element is at `list[0]`. You can access a range of items in a list by specifying the first, last, and step count. Another unique feature is *negative* indexing. `list[-1]` returns the last item in a list, `[-2]` the second to last item, etc.

A *tuple* is an ordered collection of values that is very similar to a list. The key difference is that tuples are *immutable*, meaning their contents cannot be modified after they are created. While lists are typically used for data that may change over time, tuples are often used to represent a fixed collection of related values, such as coordinates, dimensions, or configuration settings. In practice, you will encounter tuples frequently when functions return multiple values or when a programmer wants to indicate that a collection of data should not be altered.

````python
# Fixed image dimensions 
image_size = (224, 224)
print(image_size)

width, height = image_size  # Unpacking a tuple
print(f"Width: {width}, Height: {height}")

# Fixed RGB color (red)
red = (255, 0, 0)
print(red)    

# Variable RGB color (so use a list)
color = [255, 0, 0]
print(color)

# We can change color, but not red
color[1] = 255
red[1] = 255 # This will raise an error because tuples are immutable
````

While you may not find yourself creating tuples very often, you will encounter them constantly when working with scientific computing libraries such as NumPy and PyTorch. In particular, array and tensor dimensions are typically represented as tuples, making it important to recognize and understand this data structure.

### 8. Tensor Computations with Pytorch

The [tensor](https://docs.pytorch.org/tutorials/beginner/blitz/tensor_tutorial.html) is the fundamental data structure used throughout modern deep learning frameworks. Much of what we've already learned about NumPy arrays directly also applied to PyTorch tensors. Tensor operations will feel very familiar: PyTorch supports shapes, indexing, slicing, vectorized arithmetic, and multidimensional representations of data. 

The primary advantages of tensors over NumPy arrays are a) their ability to leverage hardware accelerators such as GPUs and b) their support for *automatic differentiation* (*autograd*), which allows machine learning models to automatically compute gradients during training. In many ways, a PyTorch tensor can be viewed as the deep-learning counterpart to a NumPy array:

````python
import torch

x = torch.tensor([1, 2, 3])
print(x)
print(x + 3)
print(x * 2)
````

We will convert back and forth betwen NumPy arrays and PyTorch tensors as needed throughout the semester: 

````python
# Convert x to a NumPy array:
y = x.numpy()
print(type(y))
print(y)

# Convert y to a PyTorch tensor:
z = torch.tensor(y)
print(type(z))
print(z)
# 
````

Your initial takeaway from this example might be "looks nearly identical, so what's the point of tensors?". We mentioned this above, but the key value in PyTorch is its ability to automatically calculate the derivatives of functions defined using these tensors. Since training a machine learning model involves repeatedly computing gradients and updating parameters, automating this process dramatically simplifies model development.

The following example starts by computing $y=x^3$ on a range of values using tensors. Note the separate flag `requires_grad=True`, that indicates we would like PyTorch to begin tracking operations on tensor `x`. 


````python
x = torch.tensor(2.0, requires_grad=True)
y = x**3
print(y)
````

We can calculate these derivatives using `backward()` on any scalar tensor. You will get an error if you attempt to do this on an array of tensors. While this may seem limiting, our most common use case will be for calculating the loss function that returns a single scalar value. 

In this example, we calculate $y = \mathbf{w} \cdot \mathbf{x}$ (a vector dot product), and compute the gradient with respect to $w$ and $x$:

````python
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
w = torch.tensor([2.0, 4.0, 8.0], requires_grad=True)
y = torch.dot(w, x)
print(y)

y.backward()

print(w.grad)
print(x.grad)
````

Throughout this course, we will generally trust PyTorch's automatic differentiation system to compute gradients correctly, and focus more on making predictions and computing loss functions. However, because machine learning relies heavily on matrix calculus and gradient-based optimization, it is important to develop an intuition for what autograd is doing behind the scenes. Understanding how gradients are computed will help you better interpret, debug, and reason about the models we build later in the semester.

For this previous example, let's break down what is happening in more detail. Our dot-product equation:

$$ {y} = \mathbf{w} \cdot \mathbf{x} $$

with

$$\mathbf{w} = \begin{bmatrix}
w_1 \\
w_2 \\
w_3
\end{bmatrix} \mathrm{,\quad and\quad}\mathbf{x} = \begin{bmatrix}
x_1 \\
x_2 \\
x_3
\end{bmatrix} $$

Our dot-product can be rewritten as:

$$ y = \mathbf{w}^\top \mathbf{x} = w_1 x_1 + w_2 x_2 + w_3 x_3 $$

with the partial derivatives with respect to $\mathbf{w}$ as:

$$ \frac{\partial y}{\partial w_1}
=
\frac{\partial}{\partial w_1}
\left(w_1x_1 + w_2x_2 + w_3x_3\right) = x_1$$

$$ \frac{\partial y}{\partial w_2}
=
\frac{\partial}{\partial w_2}
\left(w_1x_1 + w_2x_2 + w_3x_3\right) = x_2$$

$$ \frac{\partial y}{\partial x_3}
=
\frac{\partial}{\partial w_3}
\left(w_1x_1 + w_2x_2 + w_3x_3\right) = x_3$$


and for $\mathbf{x}$:

$$ \frac{\partial y}{\partial x_1}
=
\frac{\partial}{\partial x_1}
\left(w_1x_1 + w_2x_2 + w_3x_3\right) = w_1$$

$$ \frac{\partial y}{\partial x_2}
=
\frac{\partial}{\partial x_2}
\left(w_1x_1 + w_2x_2 + w_3x_3\right) = w_2$$

$$ \frac{\partial y}{\partial w_3}
=
\frac{\partial}{\partial x_3}
\left(w_1x_1 + w_2x_2 + w_3x_3\right) = w_3$$

Taken together, we have:

$$\nabla_w y = \mathbf{x} \mathrm{,\quad and\quad} \nabla_x y = \mathbf{w}$$

which is what we saw when we printed `w.grad()` and `x.grad()`. 


The method is called `backward()` because PyTorch computes gradients by traversing the computational graph backward, starting from the final output (often a loss function) and applying the chain rule to determine how each earlier variable contributed to that result. In other words, while the original computation proceeds forward from inputs to outputs, gradient computation proceeds backward from outputs to inputs.

We can see some evidence of this "under the hood" behavior when we print our output tensors:

````python
a = torch.tensor(2.0, requires_grad=True)
y1 = a + 2
y2 = a * 2
y3 = a ** 2
y4 = torch.sin(a)
y5 = torch.exp(a)

print(y1, y2, y3, y4, y5)
````

You can see the chain rule in action by tracing the computation all the way back to the inputs:

````python
b = torch.tensor(1.0, requires_grad=True)

z = b**2 * torch.sin(b) + b**3 * torch.exp(b)
z.backward()
print(b.grad)

# Now, go backwards through the graph (partially)
print(z.grad_fn)
print(z.grad_fn.next_functions)
print(z.grad_fn.next_functions[0][0].next_functions)
````

Properly traversing the entire computation graph is best done recursively. Here's a helper function that does a better job of visualizing the compute hierarchy:

````python
def print_computation_graph(fn, level=0):
    if fn is None:
        return

    print("  " * level + str(type(fn).__name__))

    for next_fn, _ in fn.next_functions:
        print_computation_graph(next_fn, level + 1)

print_computation_graph(z.grad_fn)
````

We can use PyTorch's automatic differentiation system on virtually any computation built from differentiable tensor operations, provided that the final result is a single scalar value. This allows PyTorch to compute gradients with respect to every element of the original tensor, making automatic differentiation practical for vectors, matrices, and the large multidimensional tensors encountered in machine learning.
